#### MODULE 2 HIGH PERFORMANCE NUMERICAL ANALYSIS (NUMPY)
Apply Vectorised operations(no slow-loops) to educational metrics using Numpy - broadcasting and masking. Produce a Vectorised efficiency report.

```python
###SETUP: We import python libraries, and then load the dataset and view the first 5 rows.
#pandas for handling tabular data, numpy for fast numerical operations, and scipy for statistical testing.

import pandas as pd
import numpy as np
from scipy import stats
us_education_data = pd.read_csv('us_education_system_dataset.csv')
us_education_data.head(5)
```

```python
#We will check for funding gaps to apply vectorised operation.
# Every school reports three funding sources — federal, state,and local. 
#Together, these should add up to roughly 100% of a school's funding, since 
#they represent the whole funding picture. If a school's total isn't close to 
#100%, that's a signal of a data quality issue (a typo or missing values)

#We will calculate the total funding percentage for all 5,000 schools at once using a vectorized operation (no loop)
#then use a boolean mask to flag any school whose total falls outside a reasonable range around 100%.

# Vectorized operation: add the three funding percentage columns for every school at once.
# This creates one new Series of 5,000 values 

total_funding_pct = us_education_data['federal_funding_pct'] + us_education_data['state_funding_pct'] + us_education_data['local_funding_pct']

# Masking: create a True/False flag for every school where the total is NOT close to 100%.
# We allow a small margin (99–101) to account for rounding in the original data.
funding_gap_mask = (total_funding_pct < 99) | (total_funding_pct > 101)

# Use the mask to pull out only the flagged (suspicious) schools into their own DataFrame.
flagged_schools = us_education_data[funding_gap_mask]

#check how many schools were flagged
print(f"The Number of Schools With Funding Gaps are As Follows: {funding_gap_mask.sum()}")
```

Zero (0) means that the funding percentage in the dataset are clean and consistent, so there are no funding gaps.


##### Comparing Loop vs. Vectorized Operation

We used NumPy/Pandas vectorized approach to calculate total funding percentage instantly. 
But to produce a vectorised efficiency report (as the brief asks), we will compare it against the loop method,
(looping through each school one row at a time doing the same calculation). We will time both methods to show the 
real difference in speed at scale, for the 5,000 schools.

```python
import time      #library used to measure the time or speed of a code execution

#Method 1: Slow loop-based approach ---
# We go row by row, manually adding the three funding percentages, one school at a time.

start_time = time.time()    #Record the start time

loop_results = []     #an empty list to store our answers as we calculate one by one.
for i in range(len(us_education_data)):  #we have 5K rows. Loop through all rows and call each number i and count from 0 to 4999.
    total = (us_education_data['federal_funding_pct'].iloc[i]  #fetch the value of the federal funding percentage for each school, 
             + us_education_data['state_funding_pct'].iloc[i]  #fetch the value of the state funding percentage for each school,
             + us_education_data['local_funding_pct'].iloc[i]) #fetch the value of the local funding percentage for each school.
    loop_results.append(total) #add the total percentage of the 3 funding sources for each school.

loop_time = time.time() - start_time    #stop the timer and calculate the time taken for the loop-based approach-
#current time minus start time = how many seconds the whole loop took to run.

print(f"Loop-based method took: {loop_time:.4f} seconds")


#Method 2: Vectorized approach ---

start_time = time.time()

vectorized_result = (us_education_data['federal_funding_pct'] 
                      + us_education_data['state_funding_pct'] 
                      + us_education_data['local_funding_pct'])

vectorized_time = time.time() - start_time
print(f"Vectorized method took: {vectorized_time:.4f} seconds")

#The Result ---

print(f"\nVectorization was approximately {loop_time / vectorized_time:.0f}x faster")
```

#### MODULE 5 - OUTLIER SURGERY (DETECT, ERASE, HEAL) 
Find outliers in an column of choice using IQR fences. convert flagged values to NaN, then replace them with the sensible group median, using fancy indexing.

```python
#STEP 1: DETECT OUTLIERS
#We will look for outliers in the per_pupil_expenditure_usd column.
#We will calculate IQR fences for per_pupil_expenditure_usd separately within each 
#nces_locale + control group, since normal spending levels differ by school type 
#and location. Any school spending outside its own group's fence is flagged as 
#a statistical outlier.


#calling the function to calculate the lower and upper IQR fences

def calculate_fences(values):
    Q1 = values.quantile(0.25)
    Q3 = values.quantile(0.75)
    IQR = Q3 - Q1
    lower_fence = Q1 - 1.5 * IQR
    upper_fence = Q3 + 1.5 * IQR
    return lower_fence, upper_fence

#Apply this to each group (locale + control), and flag any school outside its own group's fence

def flag_outliers(group):
    lower, upper = calculate_fences(group['per_pupil_expenditure_usd'])
    return (group['per_pupil_expenditure_usd'] < lower) | (group['per_pupil_expenditure_usd'] > upper)

#groupby(...).apply(...) runs flag_outliers() separately for every locale+control group,
#then stitches the True/False results back together in the original row order

outlier_mask = us_education_data.groupby(['nces_locale', 'control'], group_keys=False).apply(flag_outliers)

#Show how many schools were flagged in total

print(f"Number of schools flagged as spending outliers: {outlier_mask.sum()}")
```

```python
#STEP 2: ERASE
#We will convert the 128 flagged outliers to NaN using Boolean Indexing, not deleting the 
#whole school's row, just blanking out the suspicious numbers so they can 
#be replaced with a more sensible value in the next step.


#Make a copy first, so we don't lose the original data

us_education_data['per_pupil_expenditure_cleaned'] = us_education_data['per_pupil_expenditure_usd'].copy()

#Use the outlier_mask (True/False) we built in Step 1 to find the flagged rows,
#and set ONLY those values to NaN in our new column

us_education_data.loc[outlier_mask, 'per_pupil_expenditure_cleaned'] = np.nan

#Confirm that our 128 outliers from Step 1 are fully captured.

print(f"Values converted to NaN: {us_education_data['per_pupil_expenditure_cleaned'].isna().sum()}")
```

```python
#STEP 3: HEAL (FANCY INDEXING)
#We replace the 128 NaN values with the median per_pupil_expenditure of their own locale + control group, 
#so each replacement value reflects what's normal for that specific type of school, not the whole dataset.


#Calculate the median spending for each locale + control group, and map it back 
#to every single row (one number PER SCHOOL, matching that school's own group)

group_medians = us_education_data.groupby(['nces_locale', 'control'])['per_pupil_expenditure_usd'].transform('median')

#Use fancy indexing: outlier_mask picks out exactly the 128 flagged rows on 
#BOTH sides at once, and fills them with that row's own group median

us_education_data.loc[outlier_mask, 'per_pupil_expenditure_cleaned'] = group_medians[outlier_mask]

#check if the NaN values have been successfully replaced with the median value

print(f"Remaining NaN values: {us_education_data['per_pupil_expenditure_cleaned'].isna().sum()}")
```

#### MODULE 6 - INFERENTIAL STATISTICS AND HYPOTHESIS TESTING (SciPy)
Run a t-test(graduation rates: public vs private), a one-way ANOVA (student-teacher ratio across locales), and a bootstrap resampling (95% confidence interval on mean test score, 10,000 iterations).

```python
#Run a t-test to compare graduation rates between public and private schools
#Step 1a: To do this, we will create a simplified two-category grouping column
#Anything starting with "Public" becomes "Public", everything starting with 'private' becomes "Private"

us_education_data['school_group'] = us_education_data['control'].apply(
    lambda x: 'Public' if 'Public' in x else 'Private'
)       #look at this school's control type (x); if the word 'Public' appears anywhere in it, label it 'Public'
         #otherwise, label it 'Private'
         #this new simplified label gets stored in a brand-new column, 'school_group'


#Step 1b: Split graduation rates into two separate groups based on that new column


public_grad_rates = us_education_data[us_education_data['school_group'] == 'Public']['graduation_rate_pct']
private_grad_rates = us_education_data[us_education_data['school_group'] == 'Private']['graduation_rate_pct']
#This filters the dataset down to only rows where school_group is 'Public' then grabs just their graduation_rate_pct values 
#giving us one clean list of graduation rates, only for public schools. The next line does the same for private schools.


#Step 1c: Run the independent t-test comparing the two groups


t_statistic, p_value = stats.ttest_ind(public_grad_rates, private_grad_rates)
#takes the two separate lists and compares their averages, accounting for how spread out each group's numbers are. 
#It returns two numbers: the t_statistic (a measure of how far apart the two averages are, 
#in standardized units) and the p_value.

print(f"Public schools average graduation rate: {public_grad_rates.mean():.2f}%")
print(f"Private schools average graduation rate: {private_grad_rates.mean():.2f}%")
print(f"T-statistic: {t_statistic:.4f}")
print(f"P-value: {p_value:.4f}")



#if p_value is less than 0.05, that means the difference in graduation rates between Public and Private schools 
#is statistically significant, unlikely to be random chance, so we'll reject the Null Hypothesis (H0).
#If it's 0.05 or higher, we don't have strong enough evidence to say the difference is real, and we'd fail to reject H0.
```

##### Interpreting the T-Test Result

With a p-value of 0.0000 (far below 0.05), we reject the null hypothesis. There is a statistically significant difference in graduation rates between Public and Private schools. On average, Private schools in this dataset graduate students at 
a higher rate (87.52%) than Public schools (84.38%). The difference is very unlikely to be due to random chance.

Note: this result shows the two groups differ significantly in graduation rates. it does not by itself explain WHY. Private schools may have smaller class sizes, more selective admissions, or different student populations, which this test alone cannot confirm.

```python
#STEP 2: One way ANOVA (student-teacher ratio across locales)

#A minister wants to know if student-teacher ratios differ significantly across some locales.
#While t-test compares two groups, ANOVA (Analysis of Variance) is a statistical test that compares the means of 
#three or more groups to see if at least one group mean is different from the others. 
#In this case, we will use ANOVA to compare the student-teacher ratios across the locales.

#Our nces_locale column has 8 detailed categories (City-Large, Suburban-large, surburban-Midsize, Town-Distant, Rural-remote
#Rural-Fringe, Town-Fringe, city-midsize). 
#We simplify these into 4 broad groups - Urban, Suburban, Town, Rural - by taking the first word of each 
#category, then run a one-way ANOVA comparing student-teacher ratios across the four groups.



# Step A: Simplify the 8 detailed locale categories into 4 broad groups
# We convert nces_locale to plain text first (.astype(str)) because it's stored 
# as 'category' dtype from Module 3, which causes .str.split() to behave incorrectly
us_education_data['locale_group'] = us_education_data['nces_locale'].astype(str).str.split(' - ').str[0]

# "City" doesn't match the brief's "Urban" so we rename it so our labels match
us_education_data['locale_group'] = us_education_data['locale_group'].replace('City', 'Urban')

# Confirm the new grouping worked. it should show 4 categories now
print(us_education_data['locale_group'].value_counts())
```

```python
# Step B: Split student_teacher_ratio into four separate groups, one per locale_group

urban_ratio = us_education_data[us_education_data['locale_group'] == 'Urban']['student_teacher_ratio']
suburban_ratio = us_education_data[us_education_data['locale_group'] == 'Suburban']['student_teacher_ratio']
town_ratio = us_education_data[us_education_data['locale_group'] == 'Town']['student_teacher_ratio']
rural_ratio = us_education_data[us_education_data['locale_group'] == 'Rural']['student_teacher_ratio']


# Step C: Run the one-way ANOVA across all four groups at once
f_statistic, p_value = stats.f_oneway(urban_ratio, suburban_ratio, town_ratio, rural_ratio)

print(f"\nF-statistic: {f_statistic:.4f}")
print(f"P-value: {p_value:.4f}")
```

##### Interpreting the ANOVA Result

With an F-statistic of 0.2463 and a p-value of 0.8640 (far above 0.05), we fail to reject the null hypothesis. There is no statistically significant difference in student-teacher ratios across Urban, Suburban, Town, and Rural locales. This does not
prove the ratios are identical. It means we don't have strong enough evidence to say locale affects student-teacher ratio in this dataset.

```python
#STEP 3: BOOTSTRAP RESAMPLING (a 95% confidence interval on the mean test score, using 10,000 resampled iterations)
#We generate a 95% confidence interval for the mean math_proficiency_pct by 
#resampling our data (with replacement) 10,000 times, calculating the mean each 
#time, and taking the range that covers the middle 95% of those results

# Set a seed so our random resampling is reproducible, so anyone re-running 
# this notebook gets the exact same result, not a different random outcome

np.random.seed(42)

math_scores = us_education_data['math_proficiency_pct'].dropna()  #pull just the math_proficiency columnand drop all misssing values
                                                                   #there. there shouldnt be any but just a safety habit

# Run 10,000 resampling iterations

bootstrap_means = []   #our empty list to catch our 10,000 results
for i in range(10000):  #a loop that repeats 10000 times.
    # Draw a new sample the same size as our original, WITH replacement
    resample = np.random.choice(math_scores, size=len(math_scores), replace=True)  #randomly picks values from math_scores
    #size=len means pick the same number of values as the original dataset (5,000). 
    #replace=True means we can pick the same value more than once. 
    bootstrap_means.append(resample.mean())  #calculate the average of that one resampled version and save to the list. the loop
                                              #goes back again and does this again for 10,000 times

# Take the 2.5th and 97.5th percentiles of our 10,000 means. this is our 95% CI
lower_bound = np.percentile(bootstrap_means, 2.5)
upper_bound = np.percentile(bootstrap_means, 97.5)

print(f"Original sample mean: {math_scores.mean():.2f}%")
print(f"95% Confidence Interval: ({lower_bound:.2f}%, {upper_bound:.2f}%)")
```

##### Interpreting the Bootstrap Confidence Interval

The average math proficiency across all schools is 60.01%. Using bootstrap resampling (10,000 iterations), we are 95% confident the true mean falls between 59.55% and 60.46%. This is a narrow interval, meaning our estimate of average math proficiency is stable and reliable, not likely to shift much if we had a slightly different set of schools.

##### Poverty and Internet Access vs Graduation Rate (Module 7)
##### Understanding Our Two Key Indicators

Before presenting findings, the two socio-economic indicator columns from the dataset needs to be explained first.

1. Free/Reduced Lunch Percentage: this measures how many students at a school come from low-income households (it is the standard way US schools identify poverty levels). It does NOT measure how many students are receiving food assistance as a benefit. It measures how poor the surrounding student population is.

2. Broadband Access Percentage: this measures how many students in a school's  area have reliable home internet access. Lack of access makes it harder for students to complete assignments at home, a barrier commonly known as the "homework gap."

##### Overview

With these terms clarified, schools with a large numbers of low-income students tend to have lower graduation rates, while schools in areas with better home internet access tend to have higher graduation rates. Private schools also graduate students at a higher rate than public schools, a gap too large to be explained by chance. Locale (Urban, Suburban, 
Town, Rural) does not meaningfully affect how many teachers are available per student. That gap is not driven by geography.
